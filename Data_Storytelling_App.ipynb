```python
%%writefile app.py

import streamlit as st
import pandas as pd
import plotly.express as px

# -------------------------
# Load Dataset
# -------------------------
df = pd.read_csv("netflix.csv")

st.title("🎬 Netflix Data Storytelling App")

# -------------------------
# Dataset Introduction
# -------------------------
st.header("📖 Dataset Introduction")

movies = len(df[df["type"] == "Movie"])
tvshows = len(df[df["type"] == "TV Show"])

countries = df["country"].dropna().nunique()

col1, col2, col3 = st.columns(3)

col1.metric("Movies", movies)
col2.metric("TV Shows", tvshows)
col3.metric("Countries", countries)

# -------------------------
# EDA
# -------------------------
st.header("📊 Exploratory Data Analysis")

# Content by Year
df["release_year"] = pd.to_numeric(df["release_year"], errors="coerce")

year_count = df["release_year"].value_counts().sort_index()

st.subheader("Content by Release Year")

fig1 = px.line(
    x=year_count.index,
    y=year_count.values,
    labels={"x":"Year","y":"Count"}
)

st.plotly_chart(fig1)

# Top Countries

st.subheader("Top Countries")

country_count = (
    df["country"]
    .dropna()
    .str.split(", ")
    .explode()
    .value_counts()
    .head(10)
)

fig2 = px.bar(
    x=country_count.index,
    y=country_count.values,
    labels={"x":"Country","y":"Titles"}
)

st.plotly_chart(fig2)

# Top Genres

st.subheader("Top Genres")

genre_count = (
    df["listed_in"]
    .str.split(", ")
    .explode()
    .value_counts()
    .head(10)
)

fig3 = px.bar(
    x=genre_count.index,
    y=genre_count.values,
    labels={"x":"Genre","y":"Count"}
)

st.plotly_chart(fig3)

# Rating Distribution

st.subheader("Ratings Distribution")

rating_count = df["rating"].value_counts()

fig4 = px.pie(
    values=rating_count.values,
    names=rating_count.index
)

st.plotly_chart(fig4)

# Movies vs TV Shows

st.subheader("Movies vs TV Shows")

type_count = df["type"].value_counts()

fig5 = px.pie(
    values=type_count.values,
    names=type_count.index
)

st.plotly_chart(fig5)

# -------------------------
# Insights
# -------------------------

st.header("🔍 Key Insights")

st.write("• Most content comes from the USA.")
st.write("• Netflix content increased rapidly after 2015.")
st.write("• Drama is one of the most common genres.")
st.write("• Movies significantly outnumber TV Shows.")
st.write("• Content production has grown steadily over time.")

# -------------------------
# Recommendations
# -------------------------

st.header("💡 Recommendations")

st.write("• Increase regional language content.")
st.write("• Expand underrepresented genres.")
st.write("• Invest more in emerging markets.")
st.write("• Create more original TV Shows.")
```
